```python
from ensure import ensure_annotations
```

```python
# Ensure Annotations are working

@ensure_annotations
def add_numbers(a: int, b: int) -> int:
    return a + b

def product_numbers(a: int, b: int) -> int:
    return a * b

addition = add_numbers(5, 3)
print(addition)

# Works as expected
product = product_numbers(5, 3) 
print(product)

#  Works with strings which is not intended, 
# so we need to use ensure_annotations to enforce type checking
product = product_numbers("5", 3) 
print(product)

@ensure_annotations
def product_numbers_v1(a: int, b: int) -> int:
    return a * b

# product = product_numbers_v1("5", 3) # Throws error as expected
product = product_numbers_v1(5, 3) # Works as expected
print(product)
```
